## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

```python
# Load the dataset and define variables first to avoid NameErrors
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
visible = df[df["impressions_90d"] >= 100]

# 1. Redo Discovery A for pages with impressions_90d > 0
active_pages = df[df['impressions_90d'] > 0]
corr_active = active_pages['search_volume'].corr(active_pages['impressions_90d'])
print(f"Correlation between search_volume and impressions_90d (only for pages with impressions > 0): {corr_active:.3f}")
```

```python
# 2. Analyze which content_type has the worst mean CTR at each position tier
# Using 'visible' which is now defined in our environment
ctr_by_type_pos = visible.groupby(['position_tier', 'content_type'])['ctr'].mean().reset_index()

# Find the content_type with the minimum mean CTR per position tier
worst_ctr_types = ctr_by_type_pos.loc[ctr_by_type_pos.groupby('position_tier')['ctr'].idxmin()]
print("Worst performing content type by mean CTR at each position tier:")
print(worst_ctr_types.sort_values(by='position_tier').to_string(index=False))
```

```python
# 3. Test another belief: Does content_age_days relate to trend_direction?
# Using 'df' which is now defined in our environment
age_by_trend = df.groupby('trend_direction')['content_age_days'].agg(['mean', 'median', 'count'])
print("Content age statistics by trend direction:")
print(age_by_trend)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style for the plot
sns.set_theme(style="whitegrid")

# Define a logical order for the trend directions
trend_order = ['new', 'up', 'stable', 'flat', 'down']

# Create the box plot
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df,
    x='trend_direction',
    y='content_age_days',
    order=trend_order,
    palette='muted',
    hue='trend_direction',
    legend=False
)

plt.title('Distribution of Content Age by Trend Direction', fontsize=14, pad=15)
plt.xlabel('Trend Direction', fontsize=12)
plt.ylabel('Content Age (Days)', fontsize=12)
plt.tight_layout()
plt.show()
```
